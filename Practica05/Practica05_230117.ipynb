```python
def bar_hor(df, col, title, color, w=None, h=None, lm=0, limit=100, return_trace=False, rev=False):
    cnt_srs = df[col].value_counts()
    yy = cnt_srs.head(limit).index[::-1]
    xx = cnt_srs.head(limit).values[::-1]

    if rev:
        yy = cnt_srs.tail(limit).index[::-1]
        xx = cnt_srs.tail(limit).values[::-1]

    trace = go.Bar(y=yy, x=xx, orientation='h',
                   marker=dict(color=color))

    if return_trace:
        return trace

    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    fig = go.Figure(data=[trace], layout=layout)
    iplot(fig)


def bar_hor_noagg(x, y, title, color, w=None, h=None, lm=0, limit=100, rt=False):
    trace = go.Bar(y=x, x=y, orientation='h',
                   marker=dict(color=color))

    if rt:
        return trace

    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    fig = go.Figure(data=[trace], layout=layout)
    iplot(fig)


def bar_ver_noagg(x, y, title, color, w=None, h=None, lm=0, rt=False):
    trace = go.Bar(y=y, x=x,
                   marker=dict(color=color))

    if rt:
        return trace

    layout = dict(title=title, margin=dict(l=lm), width=w, height=h)
    fig = go.Figure(data=[trace], layout=layout)
    iplot(fig)


trace1 = bar_hor(
    donors_df,
    'Donor City',
    "Ciudades con el mayor número de donantes",
    '#c2d2ed',
    500, 400, 200,
    limit=10,
    return_trace=True
)

trace2 = bar_hor(
    donors_df,
    'Donor State',
    "Estados con el mayor número de donantes",
    '#c2d2ed',
    500, 400, 200,
    limit=10,
    return_trace=True
)

trace4 = bar_hor(
    donors_df,
    'Donor State',
    "Estados con el menor número de donantes",
    '#c2d2ed',
    600, 400, 200,
    limit=10,
    return_trace=True,
    rev=True
)

fig = tools.make_subplots(
    rows=1,
    cols=2,
    print_grid=False,
    subplot_titles=[
        'Estados con el mayor número de donantes',
        'Estados con el menor número de donantes'
    ]
)

fig.append_trace(trace2, 1, 1)
fig.append_trace(trace4, 1, 2)

fig['layout'].update(height=400, showlegend=False)
iplot(fig)
```

### Donaciones por año 
 - 3.4.1 Número de donaciones por año 
 - 3.4.2 Montos promedio de donación por año

```python
from plotly.subplots import make_subplots
import plotly.graph_objs as go
from plotly.offline import iplot

# Convertir la fecha de donación
donations_df['Donation Date'] = pd.to_datetime(donations_df['Donation Received Date'])
donations_df['Donation Year'] = donations_df['Donation Date'].dt.year
donations_df['Donation Month'] = donations_df['Donation Date'].dt.month
donations_df['Donation Day'] = donations_df['Donation Date'].dt.day
donations_df['Donation Weekday'] = donations_df['Donation Date'].dt.weekday
donations_df['Donation Hour'] = donations_df['Donation Date'].dt.hour

# Resumen por año
t1 = donations_df.groupby(['Donation Year']).agg({
    'Donation Month': 'count',
    'Donation Amount': 'mean'
}).reset_index().rename(columns={
    'Donation Month': 'Total Donations',
    'Donation Amount': 'Average Amount'
})

x = t1['Donation Year']
y1 = t1['Total Donations']
y2 = t1['Average Amount']

trace1 = go.Scatter(
    x=x[:-1],
    y=y1[:-1],
    fill='tozeroy',
    fillcolor='#fcc45f',
    mode='none'
)

trace2 = go.Scatter(
    x=x[:-1],
    y=y2[:-1],
    fill='tozeroy',
    fillcolor='#e993f9',
    mode='none'
)

fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=(
        "<b>Donaciones por Año</b>",
        "<b>Monto Promedio de Donación por Año</b>"
    )
)

fig.add_trace(trace1, row=1, col=1)
fig.add_trace(trace2, row=1, col=2)

fig.update_layout(
    height=300,
    showlegend=False
)

fig.update_yaxes(
    showgrid=False,
    zeroline=False,
    showline=False,
    ticks='',
    showticklabels=False,
    row=1,
    col=1
)

fig.update_yaxes(
    showgrid=False,
    zeroline=False,
    showline=False,
    ticks='',
    showticklabels=False,
    row=1,
    col=2
)

iplot(fig)
```

## <a id="7.8"> 7.8 ¿En qué Año y Mes, el costo del proyecto es más alto? </a>

```python
def cbp(df, col1, col2, aggcol, func, title, cs, bottom_margin=None):
    tempdf = df.groupby([col1, col2]).agg({aggcol : func}).reset_index()
    tempdf[aggcol] = tempdf[aggcol].apply(lambda x : int(x))
    tempdf = tempdf.sort_values(aggcol, ascending=False)

    sizes = list(reversed([i for i in range(10,31)]))
    intervals = int(len(tempdf) / len(sizes))
    size_array = [9]*len(tempdf)
    
    st = 0
    for i, size in enumerate(sizes):
        for j in range(st, st+intervals):
            size_array[j] = size 
        st = st+intervals
    tempdf['size_n'] = size_array

    cols = list(tempdf['size_n'])

    trace1 = go.Scatter( x=tempdf[col1], y=tempdf[col2], mode='markers', text=tempdf[aggcol],
        marker=dict( size=tempdf.size_n, color=cols, colorscale=cs ))
    data = [trace1]
    
    # Agregar configuración de ejes
    layout = go.Layout(
        title=title,
        xaxis=dict(
            tickmode='linear',
            tick0=0,
            dtick=2
        ),
        yaxis=dict(
            tickmode='array',
            tickvals=[2013, 2014, 2015, 2016, 2017, 2018]
        ),
        margin=dict(b=150) if bottom_margin else {}
    )
    
    fig = go.Figure(data=data, layout=layout)
    iplot(fig)

# ✅ CORREGIDO: 'cost' → 'Project Cost'
cbp(projects_df, 'Posted Month', 'Posted Year', 'Project Cost', 'mean', 'Costo Promedio por Año y Mes', 'Picnic')
```

```python
import nltk
from nltk.corpus import stopwords
import string

# Intentar descargar con ruta específica o usar una lista manual
try:
    stopwords_eng = stopwords.words("english")
except LookupError:
    # Si no encuentra stopwords, usamos una lista básica en inglés
    stopwords_eng = ['i', 'me', 'my', 'myself', 'we', 'our', 'ours', 'ourselves', 'you', "you're", 
                     "you've", 'your', 'yours', 'yourself', 'yourselves', 'he', 'him', 'his', 
                     'himself', 'she', "she's", 'her', 'hers', 'herself', 'it', "it's", 'its', 
                     'itself', 'they', 'them', 'their', 'theirs', 'themselves', 'what', 'which', 
                     'who', 'whom', 'this', 'that', "that'll", 'these', 'those', 'am', 'is', 'are', 
                     'was', 'were', 'be', 'been', 'being', 'have', 'has', 'had', 'having', 'do', 
                     'does', 'did', 'doing', 'a', 'an', 'the', 'and', 'but', 'if', 'or', 'because', 
                     'as', 'until', 'while', 'of', 'at', 'by', 'for', 'with', 'about', 'against', 
                     'between', 'into', 'through', 'during', 'before', 'after', 'above', 'below', 
                     'to', 'from', 'up', 'down', 'in', 'out', 'on', 'off', 'over', 'under', 'again', 
                     'further', 'then', 'once', 'here', 'there', 'when', 'where', 'why', 'how', 'all', 
                     'any', 'both', 'each', 'few', 'more', 'most', 'other', 'some', 'such', 'no', 'nor', 
                     'not', 'only', 'own', 'same', 'so', 'than', 'too', 'very', 's', 't', 'can', 'will', 
                     'just', 'don', "don't", 'should', "should've", 'now', 'd', 'll', 'm', 'o', 're', 
                     've', 'y', 'ain', 'aren', "aren't", 'couldn', "couldn't", 'didn', "didn't", 'doesn', 
                     "doesn't", 'hadn', "hadn't", 'hasn', "hasn't", 'haven', "haven't", 'isn', "isn't", 
                     'ma', 'mightn', "mightn't", 'mustn', "mustn't", 'needn', "needn't", 'shan', "shan't", 
                     'shouldn', "shouldn't", 'wasn', "wasn't", 'weren', "weren't", 'won', "won't", 'wouldn', 
                     "wouldn't", 'the', 'a', 'an']

def _clean(txt):
    # Convertir a minúsculas
    txt = txt.lower()

    # Eliminación de signos de puntuación 
    txt = ''.join(x for x in txt if x not in string.punctuation)

    # Eliminación de stopwords (palabras vacías)
    clean_txt = ""
    for word in txt.split():
        if word in stopwords_eng:
            continue 
        clean_txt += " "
        clean_txt += word 
    
    # Eliminación de palabras de ruido adicionales
    noise = ['students','my','title', 'would', 'every', 'will', 'many','also','donotremoveessaydivider']
    for ns in noise:
        clean_txt = clean_txt.replace(ns, "")

    # Eliminar espacios en blanco al inicio y al final
    return clean_txt.strip()

# Merge usando los nombres correctos de las columnas
projects_df = projects_df.merge(schools_df, left_on='ID de la Escuela', right_on='School ID', how='inner')

# Filtrar proyectos de abril de 2018
april18_df = projects_df[(projects_df['Posted Year'] == 2018) & (projects_df['Posted Month'] == 4)]

# Aplicar la función de limpieza a las columnas de texto
april18_df['ensayo_limpio'] = april18_df['Project Essay'].apply(_clean)
april18_df['necesidad_limpia'] = april18_df['Project Need Statement'].apply(_clean)
```

```python
# Importar librerías necesarias
from wordcloud import WordCloud
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image, ImageDraw
import random
from matplotlib.colors import LinearSegmentedColormap

# Unir todos los títulos de proyectos en texto
# df1 = donante único, df2 = múltiples donantes
pos_text_cln = " ".join(df2['Project Title'])
neg_text_cln = " ".join(df1['Project Title'])

# Eliminar palabras "ruido"
noise = ['Learning', 'Student', 'Classroom', 'Need']
for n in noise:
    pos_text_cln = pos_text_cln.replace(n, " ")
    neg_text_cln = neg_text_cln.replace(n, " ")

def green_color(word, font_size, position, orientation, random_state=None, **kwargs):
    """Función para colorear palabras en tonos verdes"""
    return 'hsl({:d}, 80%, {:d}%)'.format(random.randint(85, 140), random.randint(60, 80))

def red_color(word, font_size, position, orientation, random_state=None, **kwargs):
    """Función para colorear palabras en tonos rojos"""
    return 'hsl({:d}, 80%, {:d}%)'.format(random.randint(0, 35), random.randint(60, 80))

# Crear la máscara de la silueta de persona leyendo un libro
def create_person_reading_mask(width=1000, height=1000):
    """Crear máscara con forma de persona sentada leyendo un libro"""
    mask = Image.new('L', (width, height), 255)  # Fondo blanco
    draw = ImageDraw.Draw(mask)
    
    # Cabeza (círculo en la parte superior)
    head_center_x = width // 2
    head_center_y = height // 8
    head_radius = width // 10
    draw.ellipse(
        [head_center_x - head_radius, head_center_y - head_radius,
         head_center_x + head_radius, head_center_y + head_radius],
        fill=0
    )
    
    # Cuerpo/Torso (forma de trapecio)
    body_top = head_center_y + head_radius
    body_bottom = height // 2 + height // 6
    body_top_width = width // 5
    body_bottom_width = width // 2
    
    body_points = [
        (head_center_x - body_top_width, body_top),
        (head_center_x + body_top_width, body_top),
        (head_center_x + body_bottom_width, body_bottom),
        (head_center_x - body_bottom_width, body_bottom)
    ]
    draw.polygon(body_points, fill=0)
    
    # Libro abierto (rectángulo en la parte inferior)
    book_top = body_bottom
    book_bottom = height - height // 8
    book_left = width // 5
    book_right = width - width // 5
    
    draw.rectangle([book_left, book_top, book_right, book_bottom], fill=0)
    
    # Brazos
    arm_width = width // 15
    
    # Brazo izquierdo
    draw.polygon([
        (head_center_x - body_top_width, body_top + (body_bottom - body_top) // 3),
        (head_center_x - body_top_width + arm_width * 2, body_top + (body_bottom - body_top) // 3),
        (book_left + arm_width * 2, book_top + (book_bottom - book_top) // 3),
        (book_left, book_top + (book_bottom - book_top) // 3)
    ], fill=0)
    
    # Brazo derecho
    draw.polygon([
        (head_center_x + body_top_width - arm_width * 2, body_top + (body_bottom - body_top) // 3),
        (head_center_x + body_top_width, body_top + (body_bottom - body_top) // 3),
        (book_right, book_top + (book_bottom - book_top) // 3),
        (book_right - arm_width * 2, book_top + (book_bottom - book_top) // 3)
    ], fill=0)
    
    return np.array(mask)

# Crear la máscara
mask = create_person_reading_mask(1000, 1000)

# Crear colormap personalizado para Single Donors (tonos cálidos suaves - similar a la referencia)
colors_single = [
    (0.95, 0.75, 0.70),  # Rosa claro
    (0.90, 0.60, 0.55),  # Salmón
    (0.85, 0.45, 0.50),  # Rosa medio
    (0.75, 0.35, 0.55),  # Rosa oscuro
    (0.65, 0.30, 0.60),  # Morado claro
    (0.55, 0.25, 0.55),  # Morado
    (0.80, 0.50, 0.40),  # Naranja suave
    (0.70, 0.40, 0.35),  # Terracota
]
custom_cmap_single = LinearSegmentedColormap.from_list('custom_single', colors_single)

# Crear la figura con 2 subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=[18, 9], facecolor='#fef7ea')

# Nube de palabras para proyectos con Donante Único (colores cálidos suaves)
wordcloud1 = WordCloud(
    max_words=3000,
    colormap=custom_cmap_single,
    background_color='rgb(254, 247, 234)',
    mask=mask,
    contour_width=0,
    min_font_size=5,
    max_font_size=75,  # Aumentado para que palabras como "Technology" y "Reading" dominen más
    prefer_horizontal=0.85,
    relative_scaling=0.3,  # Ligeramente mayor para que las palabras más frecuentes sean más grandes
    scale=3,
    collocations=False,
    random_state=42
).generate(neg_text_cln)

ax1.imshow(wordcloud1, interpolation='bilinear')
ax1.axis('off')
ax1.set_title('TopWords - Proyectos con Donante Único', fontsize=14, pad=20)

# Nube de palabras para proyectos con Múltiples Donantes (colores variados)
wordcloud2 = WordCloud(
    max_words=3000,
    colormap='nipy_spectral',
    background_color='rgb(254, 247, 234)',
    mask=mask,
    contour_width=0,
    min_font_size=5,
    max_font_size=75,  # Mismo tamaño para consistencia
    prefer_horizontal=0.85,
    relative_scaling=0.3,
    scale=3,
    collocations=False,
    random_state=42
).generate(pos_text_cln)

ax2.imshow(wordcloud2, interpolation='bilinear')
ax2.axis('off')
ax2.set_title('TopWords - Proyectos con Múltiples Donantes', fontsize=14, pad=20)

plt.tight_layout()
plt.show()
```
